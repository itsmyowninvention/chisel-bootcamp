# 理解Scala
Scala是另一种支持常见编程范式的编程语言。我们选择使用它有几个原因：

- 它是托管嵌入式DSL的良好语言。

# 条件语句

Scala像其他编程语言一样实现条件语句。

````scala
// A simple conditional; by the way, this is a comment
if (numberOfKittens > kittensPerHouse) { 
    println("Too many kittens!!!") 
}
// The braces are not required when all branches are one liners. However, the 
// Scala Style Guide prefers brace omission only if an "else" clause is included.
// (Preferably not this, even though it compiles...)
if (numberOfKittens > kittensPerHouse) 
    println("Too many kittens!!!")

// ifs have else clauses, of course
// This is where you can omit braces!
if (done) 
    println("we are done")
else 
    numberOfKittens += 1

// And else ifs
// For style, keep braces because not all branches are one liners. 
if (done) {
    println("we are done")
}
else if (numberOfKittens < kittensPerHouse) {
    println("more kittens!")
    numberOfKittens += 1
}
else {
    done = true
}
````

但在Scala中，"`if`" 条件语句会返回一个值。这个值是什么？它由所选分支的最后一行给出。这非常强大，特别是在用于初始化函数和类中的值时。它看起来像这样：

````scala
val likelyCharactersSet = if (alphabet.length == 26)
    "english"
else 
    "not english"

println(likelyCharactersSet)
````

我们创建了一个常量 *likelyCharactersSet*，其值在运行时由条件确定。

# 方法（函数）

方法使用关键字 `def` 定义。在这里，我们将滥用符号，也将它们称为函数。函数参数（或参数）在逗号分隔的列表中指定，该列表指定参数的名称、类型以及可选的默认值。为了清晰起见，应指定返回类型。

没有任何参数的Scala函数不需要空括号。这通常使开发人员的生活更轻松，在类成员成为函数的情况下，因为引用它涉及一些计算。按照惯例，没有副作用（即调用它们不会改变任何东西，它们只是返回一个值）的无参数函数不使用括号，而有副作用（可能它们改变类变量或打印内容）的函数应该需要括号。

## 简单声明

````scala
// Simple scaling function with an input argument, e.g., times2(3) returns 6
// Curly braces can be omitted for short one-line functions.
def times2(x: Int): Int = 2 * x

// More complicated function
def distance(x: Int, y: Int, returnPositive: Boolean): Int = {
    val xy = x * y
    if (returnPositive) xy.abs else -xy.abs
}
````

## 重载函数
相同的函数名可以以多种方式使用。参数及其类型确定一个签名，允许编译器确定应调用哪个版本的函数。不过，应避免使用重载函数。

````scala
// Overloaded function
def times2(x: Int): Int = 2 * x
def times2(x: String): Int = 2 * x.toInt

times2(5)
times2("7")
````

## 递归和嵌套函数
花括号定义代码作用域。在函数的作用域内可能存在更多函数或递归函数调用。在特定作用域中定义的函数只能在该作用域内访问。

````scala
/** Prints a triangle made of "X"s
  * This is another style of comment
  */
def asciiTriangle(rows: Int) {
    
    // This is cute: multiplying "X" makes a string with many copies of "X"
    def printRow(columns: Int): Unit = println("X" * columns)
    
    if(rows > 0) {
        printRow(rows)
        asciiTriangle(rows - 1) // Here is the recursive call
    }
}

// printRow(1) // This would not work, since we're calling printRow outside its scope
asciiTriangle(6)
````

# 列表
Scala实现了各种聚合或序列对象。列表很像数组，但支持附加和提取的额外操作。

````scala
val x = 7
val y = 14
val list1 = List(1, 2, 3)
val list2 = x :: y :: y :: Nil       // An alternate notation for assembling a list

val list3 = list1 ++ list2           // Appends the second list to the first list
val m = list2.length
val s = list2.size

val headOfList = list1.head          // Gets the first element of the list
val restOfList = list1.tail          // Get a new list with first element removed

val third = list1(2)                 // Gets the third element of a list (0-indexed)
````

# `for` 语句

Scala有一个 `for` 语句，它像传统的for语句一样工作。您可以迭代一系列值。

````scala
for (i <- 0 to 7) { print(i + " ") }
println()
````

使用 `until` 而不是 `to` 来从0迭代到6（不包括7）。

````scala
for (i <- 0 until 7) { print(i + " ") }
println()
````

添加一个 `by` 以按固定量递增。以下打印出0到10之间的偶数整数。

````scala
for(i <- 0 to 10 by 2) { print(i + " ") }
println()
````

如果您有某种集合并想访问所有元素，可以像Java和Python中一样使用 `for` 作为*迭代器*。这里我们创建一个包含4个随机整数的列表，然后对它们求和。

````scala
val randomList = List(scala.util.Random.nextInt(), scala.util.Random.nextInt(), scala.util.Random.nextInt(), scala.util.Random.nextInt())
var listSum = 0
for (value <- randomList) {
  listSum += value
}
println("sum is " + listSum)
````

Scala的 `for` 还有很多技巧。它将直观地适用于广泛的传统迭代需求，但它可能不是最方便使用的工具。像对数组元素求和这样的操作通常更容易使用称为*推导式*的丰富函数族来完成，这些函数在许多不同的元素集合中都可用。后面的模块将提供更多关于 `for` 及其相关功能的信息。

# Scala是一种面向对象的语言
Scala是面向对象的，理解这一点很重要，以便充分利用Scala和Chisel。请注意，毫无疑问，有多种方式可以描述所有这些。
1. 变量是对象。
1. 在*Scala*的 `val` 声明意义上的常量也是对象。
1. 甚至字面值也是对象。
1. 甚至函数本身也是对象。稍后会详细介绍。
1. 对象是类的实例。
  1. 事实上，在Scala中几乎所有重要的方面，*面向对象*中的*对象*都将被称为实例。
1. 在定义类时，程序员指定：
  1. 与类关联的数据（`val`、`var`）。
  1. 类实例可以执行的操作，称为方法或函数。
1. 类可以扩展其他类。
  1. 被扩展的类是超类；扩展者是子类。
  1. 在这种情况下，子类继承超类的数据和方法。
  1. 有许多有用但受控的方式，类可以扩展或覆盖继承的属性。
1. 类可以从特质继承。将特质视为轻量级类，允许从多个超类继承的特定、有限方式。
1. （单例）对象是一种特殊的Scala类。
  1. 它们不是上面提到的对象。记住，我们称那些为实例。

我们即将了解如何在Scala中创建类。

# 类示例
创建Scala类的示例可能是

````scala
// WrapCounter counts up to a max value based on a bit size
class WrapCounter(counterBits: Int) {

  val max: Long = (1 << counterBits) - 1
  var counter = 0L
    
  def inc(): Long = {
    counter = counter + 1
    if (counter > max) {
        counter = 0
    }
    counter
  }
  println(s"counter created with max value $max")
}
````

这里有什么：
* ```class WrapCounter``` -- 这是 **WrapCounter** 的定义。
* ```(counterBits: Int)``` -- 创建WrapCounter需要一个整数参数，命名得很好，表明它是计数器的位宽。
* 花括号（{}）界定一个代码块。大多数类使用代码块来定义变量、常量和方法（函数）。
* `val max: Long =` -- 类包含一个成员变量 **max**，声明为 `Long` 类型，并在创建类时初始化。
* `(1 << counterBits) - 1` 计算可以包含在 **counterBits** 位中的最大值。由于 **max** 是用 `val` 创建的，它不能被更改。
* 一个变量 **counter** 被创建并初始化为 **0L**。**L** 表示0是一个长整型值；因此，**counter** 被推断为Long类型。
* **max** 和 **counter** 通常被称为类的_成员变量_。
* 定义了一个类方法 **inc**，它不接受参数并返回一个 **Long** 值。
* 方法 **inc** 的主体是一个代码块，其中：
  * `counter = counter + 1` 递增 **counter**。
  * `if (counter > max) { counter = 0 }` 测试计数器是否大于 **max** 值，如果是，则将其重置为零。
  * `counter` -- 代码块的最后一行很重要。
    * 任何表示为代码块最后一行的值都被视为该代码块的返回值。返回值可以被调用语句使用或忽略。
    * 这适用于相当普遍的情况；例如，由于 `if` then `else` 语句用代码块定义其真和假子句，它可以返回一个值，即 `val result = if (10 * 10 > 90) "greater" else "lesser"` 将创建一个值为 "greater" 的 `val`。
  * 所以在这种情况下，函数 **inc** 返回 **counter** 的值。
* `println(s"counter created with max value $max")` 将字符串打印到标准输出。因为 **println** 直接在定义代码块中，所以它是类初始化代码的一部分，每次创建此类的实例时都会运行，即打印出字符串。
* 在这种情况下打印的字符串是一个_插值_字符串。
  * 第一个双引号前的 **s** 标识这是一个插值字符串。
  * 插值字符串在运行时处理。
  * **\$max** 被替换为max的值。
  * 如果 **\$** 后跟一个代码块，该代码块中可以包含任意Scala代码。
    * 例如，**`println(s"doubled max is ${max + max}")`**。
    * 此代码块的返回值将替换 `${...}`。
    * 如果返回值不是字符串，它将被转换为字符串；实际上，scala中的每个类或类型都有一个隐式转换为字符串的定义。
  * 通常应避免每次创建类的实例时都打印某些内容，以免淹没标准输出，除非您正在调试。

# 代码块
代码块由花括号界定。一个块可以包含零行或多行Scala代码。Scala代码的最后一行成为代码块的返回值（可能被忽略）。没有行的代码块将返回一个特殊的类似null的对象，称为 `Unit`。代码块在整个Scala中使用：它们是类定义的主体，它们构成函数和方法定义，它们是 `if` 语句的子句，它们是 `for` 和许多其他Scala运算符的主体。

## 参数化代码块
代码块可以接受参数。在类和方法定义的情况下，这些参数看起来像大多数传统编程语言中的参数。在下面的示例中，`c` 和 `s` 是代码块的参数。

````scala
// A one-line code block doesn't need to be enclosed in {}
def add1(c: Int): Int = c + 1

class RepeatString(s: String) {
  val repeatedString = s + s
}
````

**重要**：还有另一种方式可以使代码块参数化。这是一个示例。

````scala
val intList = List(1, 2, 3)
val stringList = intList.map { i =>
  i.toString
}
````

代码块被传递给List类的 `map` 方法。`map` 方法要求其代码块具有单个参数。为列表的每个成员调用代码块，代码块返回该成员转换为字符串的结果。Scala几乎过度接受这种语法的变体。您可能会看到以许多不同方式编写的此代码。这种类型的代码块称为匿名函数，有关匿名函数的更多详细信息将在后面的模块中提供。

这里的目的是帮助您在遇到它们时识别不同的符号类型。随着您使用Scala，这些将变得更加舒适和熟悉。作者倾向于采用特定的风格，并且也存在个别语法情况，其中一种符号看起来更自然。单行代码倾向于使用更简洁的形式。复杂的块通常具有更叙述性的外观。为了使协作更容易，应考虑浏览[Scala风格指南](http://docs.scala-lang.org/style/)中的最佳实践。

# 命名参数和参数默认值
考虑以下方法定义。
```scala
def myMethod(count: Int, wrap: Boolean, wrapValue: Int = 24): Unit = { ... }
```
调用方法时，您经常会看到参数名称以及传入的值。
```scala
myMethod(count = 10, wrap = false, wrapValue = 23)
```
使用命名参数，您甚至可以使用不同的顺序调用函数。
```scala
myMethod(wrapValue = 23, wrap = false, count = 10)
```
对于频繁调用的方法，参数顺序可能很明显。但对于不太常见的方法，特别是布尔参数，在调用中包含名称可以使您的代码更具可读性。如果方法有相同类型的参数长列表，使用名称也可以减少出错的机会。类定义的参数也使用此命名参数方案（它们实际上只是类构造函数方法的参数）。

当某些参数具有默认值（不需要覆盖）时，调用者只需传递（按名称）不使用默认值的特定参数。请注意，参数 `wrapValue` 的默认值为24。因此，

```scala
myMethod(wrap = false, count = 10)
```
将像传入24一样工作。
